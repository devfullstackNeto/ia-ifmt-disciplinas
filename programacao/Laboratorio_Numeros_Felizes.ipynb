# Laboratório: **Números Felizes**

## Objetivo

Neste laboratório, vamos resolver um problema clássico de lógica de programação: **descobrir se um número é feliz**.

Mais importante do que apenas chegar ao código final, vamos construir a solução em etapas:

1. entender o problema;
2. criar uma primeira solução;
3. descobrir uma falha importante nessa solução;
4. pensar em uma condição de parada;
5. construir a solução final.

> **Ideia central:** antes de programar, precisamos entender como o problema se comporta.

## 1. O que é um número feliz?

Um número é chamado de **feliz** quando realizamos repetidamente o seguinte processo:

1. separamos seus dígitos;
2. elevamos cada dígito ao quadrado;
3. somamos os resultados;
4. repetimos o processo com o novo número.

Se em algum momento chegarmos ao número **1**, o número original é considerado **feliz**.

### Exemplo: 19

$$1^2 + 9^2 = 82$$

$$8^2 + 2^2 = 68$$

$$6^2 + 8^2 = 100$$

$$1^2 + 0^2 + 0^2 = 1$$

Portanto, **19 é um número feliz**.

## 2. Como transformar essa ideia em algoritmo?

Para cada número, precisamos calcular a **soma dos quadrados de seus dígitos**.

Por exemplo, para o número `82`:

- dígito `8` → $8^2 = 64$
- dígito `2` → $2^2 = 4$
- soma → `64 + 4 = 68`

Mas, antes de calcular essa soma, surge uma questão importante:

> **Como podemos separar os dígitos de um número?**

Vamos fazer isso de duas formas: primeiro **matematicamente**, usando divisão inteira e resto da divisão; depois usando a conversão para `str()`.

### 2.1. Separando os dígitos matematicamente

Podemos separar os dígitos de um número usando apenas operações matemáticas.

Para isso, utilizamos dois operadores importantes:

- `% 10` → obtém o **último dígito** do número;
- `// 10` → remove o **último dígito** do número.

Considere o número `82`.

Primeiro:

$$82 \% 10 = 2$$

Portanto, o último dígito é `2`.

Agora removemos esse último dígito:

$$82 // 10 = 8$$

Repetimos o processo com o número `8`:

$$8 \% 10 = 8$$

$$8 // 10 = 0$$

Quando o valor chega a `0`, terminamos de percorrer os dígitos.

Assim, a decomposição acontece desta forma:

```text
82
↓
82 % 10 = 2      ← pega o último dígito
82 // 10 = 8     ← remove o último dígito

8
↓
8 % 10 = 8
8 // 10 = 0      ← terminou
```

> Perceba que os dígitos são encontrados **da direita para a esquerda**.

````python
numero = 82
temporario = numero
soma = 0

while temporario > 0:
    # Pega o último dígito
    digito = temporario % 10
    print("Dígito encontrado:", digito)

    # Soma o quadrado do dígito
    soma = soma + digito ** 2

    # Remove o último dígito
    temporario = temporario // 10

print("Número original:", numero)
print("Soma dos quadrados dos dígitos:", soma)
````

### 2.2. Separando os dígitos com `str()`

Outra possibilidade é converter temporariamente o número para texto.

Por exemplo:

```python
str(82)
```

produz o texto:

```text
"82"
```

Como uma `string` pode ser percorrida com `for`, conseguimos acessar cada dígito separadamente.

Depois, convertemos cada caractere novamente para inteiro usando `int()`.

````python
numero = 82
soma = 0

for digito in str(numero):
    soma = soma + int(digito) ** 2

print("Número original:", numero)
print("Soma dos quadrados dos dígitos:", soma)
````

As duas estratégias resolvem o mesmo problema.

**Matematicamente:**

```python
digito = temporario % 10
temporario = temporario // 10
```

**Com conversão para texto:**

```python
for digito in str(numero):
```

A primeira abordagem ajuda a compreender a decomposição matemática de um número. A segunda é mais curta e será utilizada inicialmente nas próximas etapas.

## 3. Primeira tentativa

Agora podemos repetir esse cálculo até o número chegar a `1`.

Uma primeira solução poderia ser:

````python
numero = 19

while numero != 1:
    soma = 0

    for digito in str(numero):
        soma = soma + int(digito) ** 2

    numero = soma
    print(numero)

print("É um número feliz!")
````

### Testando com 19

A sequência será:

```text
19 → 82 → 68 → 100 → 1
```

Funcionou.

Mas será que esse algoritmo funciona para **qualquer número**?

# 4. O problema da condição de parada

Observe nossa condição:

```python
while numero != 1:
```

Ela diz:

> Continue executando enquanto o número não for 1.

Mas surge uma pergunta importante:

## E se o número nunca chegar a 1?

Vamos analisar o número **4**.

Aplicando o mesmo processo:

```text
4
↓
16
↓
37
↓
58
↓
89
↓
145
↓
42
↓
20
↓
4
```

Perceba que voltamos ao **4**.

A partir daí, toda a sequência se repetirá:

```text
4 → 16 → 37 → 58 → 89 → 145 → 42 → 20 → 4 → ...
```

Ou seja, entramos em um **ciclo infinito**.

## Problema lógico

Nossa primeira solução sabe reconhecer quando encontramos o `1`.

Porém, ela não sabe reconhecer quando começamos a repetir valores.

Precisamos então de uma segunda condição de parada:

> **Se um número aparecer novamente, significa que entramos em um ciclo.**

Então precisamos guardar os números que já apareceram.

## 5. Primeira solução para o ciclo: guardar os valores em uma lista

Se precisamos descobrir se um número já apareceu antes, uma primeira ideia bastante natural é **guardar os valores calculados**.

Podemos começar usando uma **lista**.

```python
visitados = []
```

Sempre que encontramos um novo valor, adicionamos esse valor à lista com `append()`.

````python
visitados = []

visitados.append(4)
visitados.append(16)
visitados.append(37)

print("Valores guardados:", visitados)

print("16 já apareceu?", 16 in visitados)
print("89 já apareceu?", 89 in visitados)
````

Agora conseguimos fazer uma pergunta antes de continuar:

```python
numero in visitados
```

Se o resultado for `True`, aquele valor já apareceu anteriormente.

Isso significa que a sequência começou a se repetir e entramos em um **ciclo**.

Portanto, podemos acrescentar uma segunda condição ao `while`:

```python
while numero != 1 and numero not in visitados:
```

Vamos testar primeiro usando a lista.

````python
numero = 4
visitados = []

while numero != 1 and numero not in visitados:
    # Guarda o número atual na lista
    visitados.append(numero)

    soma = 0

    for digito in str(numero):
        soma = soma + int(digito) ** 2

    print(numero, "→", soma)
    numero = soma

if numero == 1:
    print("É um número feliz!")
else:
    print("O número", numero, "já apareceu antes.")
    print("Foi detectado um ciclo.")
````

### 5.1. Melhorando a estrutura: de `list` para `set`

A lista resolveu nosso problema. Entretanto, observe o que realmente precisamos fazer durante o algoritmo:

> **Guardar valores e perguntar repetidamente se determinado valor já apareceu.**

Para esse tipo de situação, o Python possui uma estrutura muito apropriada: o **conjunto (`set`)**.

Um `set` armazena valores **sem repetição** e é especialmente adequado quando queremos testar se um elemento pertence ou não a uma coleção.

Com lista utilizávamos:

```python
visitados = []
visitados.append(numero)
```

Com conjunto utilizamos:

```python
visitados = set()
visitados.add(numero)
```

````python
visitados = set()

visitados.add(4)
visitados.add(16)
visitados.add(37)

print("Valores guardados:", visitados)

print("16 já apareceu?", 16 in visitados)
print("89 já apareceu?", 89 in visitados)
````

A lógica da condição de parada continua exatamente a mesma:

```python
while numero != 1 and numero not in visitados:
```

Agora temos duas possíveis razões para o algoritmo parar:

1. `numero == 1` → encontramos a condição de felicidade;
2. `numero in visitados` → encontramos um valor repetido e detectamos um ciclo.

Na solução final, utilizaremos `set()`.

# 6. Solução final

Agora podemos construir o algoritmo completo.

````python
numero = int(input("Digite um número inteiro positivo: "))

original = numero
visitados = set()

while numero != 1 and numero not in visitados:

    # Guarda o número atual antes de calcular o próximo
    visitados.add(numero)

    soma = 0

    # Percorre cada dígito do número
    for digito in str(numero):
        soma = soma + int(digito) ** 2

    print(numero, "→", soma)

    numero = soma

if numero == 1:
    print(f"\n{original} é um número feliz! 😄")
else:
    print(f"\n{original} não é um número feliz.")
    print("Foi detectado um ciclo de repetição.")
````

## 7. Entendendo a condição principal

A parte mais importante do algoritmo é:

```python
while numero != 1 and numero not in visitados:
```

Ela possui duas condições:

| Condição | Significado |
|---|---|
| `numero != 1` | ainda não encontramos a condição de felicidade |
| `numero not in visitados` | ainda não detectamos um ciclo |

O `while` para quando **uma dessas condições deixa de ser verdadeira**.

Depois verificamos por qual motivo ele terminou:

```python
if numero == 1:
    print("É feliz!")
else:
    print("Entrou em ciclo.")
```
