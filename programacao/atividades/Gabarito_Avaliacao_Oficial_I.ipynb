```python
n = int(input("Digite um número com 4 dígitos: "))

# verificar se tem 4 dígitos
aux = n
digitos = 0
while aux != 0:
  aux //= 10
  digitos+=1

rodadas = 0

while n != 6174 and n % 1111 != 0:

  d1 = n // 1000
  d2 = n // 100 % 10
  d3 = n // 10 % 10
  d4 = n % 10

  if d1 > d2:
    d1,d2 = d2,d1
  if d1 > d3:
    d1,d3 = d3,d1
  if d1 > d4:
    d1,d4 = d4,d1
  if d2 > d3:
    d2,d3 = d3,d2
  if d2 > d4:
    d2,d4 = d4,d2
  if d3 > d4:
    d3,d4 = d4,d3

  menor = (d1 * 1000 + d2 * 100 + d3 * 10 + d4)
  maior = (d4 * 1000 + d3 * 100 + d2 * 10 + d1)

  n = maior - menor
  rodadas+=1

  print(f"{rodadas} | {maior} - {menor} = {n}")

print(f"Total: {rodadas} rodadas.")
```
