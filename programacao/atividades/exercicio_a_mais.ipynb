## Exercício – Refatorando o Desafio de Kaprekar

Na atividade anterior, você desenvolveu um programa para analisar números de quatro dígitos e identificar quantas rodadas eram necessárias para chegar à constante de Kaprekar **6174**.

Agora, o objetivo é **refazer essa solução utilizando os novos conteúdos estudados em aula**.

### Nova versão do programa

Seu programa deverá:

1. Gerar uma **lista com 20 números aleatórios** entre `1000` e `9999`.

2. Utilizar **compreensão de lista** para criar uma nova lista contendo apenas os números válidos para o processo de Kaprekar.

   Números com os quatro dígitos iguais, como:

   `1111`, `2222`, `3333`, ...

   deverão ser descartados.

3. Para cada número válido, executar o processo de Kaprekar até chegar ao número:

**6174**

Em cada rodada:

- separe os quatro dígitos;
- coloque os dígitos em ordem crescente;
- forme o menor número;
- forme o maior número utilizando a ordem inversa;
- calcule:

**maior − menor**

- utilize o resultado na próxima rodada.

4. Armazene em uma **lista** a quantidade de rodadas necessária para cada número chegar a `6174`.

5. Descubra a **maior quantidade de rodadas** encontrada.

6. Utilize **compreensão de lista** para gerar uma lista contendo todos os números que atingiram essa quantidade máxima de rodadas.

7. Utilize pelo menos um **operador ternário** para apresentar uma mensagem diferente dependendo do resultado.

Por exemplo:

- caso apenas um número tenha atingido o máximo;
- caso vários números tenham empatado.

### Exemplo de saída

```text
Números sorteados:
[3524, 1111, 9831, 7261, 4444, 2113, ...]

Números válidos:
[3524, 9831, 7261, 2113, ...]

Rodadas:
[3, 4, 7, 5, ...]

Maior quantidade de rodadas: 7

Número(s) com maior quantidade de rodadas:
[7261]

Resultado:
Apenas um número atingiu a maior quantidade de rodadas.
```

Caso exista empate:

```text
Número(s) com maior quantidade de rodadas:
[7261, 9832, 8532]

Resultado:
3 números atingiram a maior quantidade de rodadas.
```

### Requisitos obrigatórios

Nesta nova versão, deverão ser utilizados obrigatoriamente:

- `random`;
- listas;
- **compreensão de listas**;
- **operador ternário**;
- `for`;
- `while`;
- estruturas condicionais.
