## Missão 2 · Crie a memória de pares de palavras

Agora vamos montar um dicionário. Para cada palavra, ele guarda uma lista das palavras que vieram logo depois dela no nosso corpus.

Por exemplo: `"intelligence": ["can", "can", "helps", ...]`. O `can` aparece mais de uma vez porque foi encontrado mais de uma vez nas frases. **Não vamos eliminar essas repetições:** elas serão importantes quando começarmos a contar frequências.

Essa é uma maneira simples de representar relações entre duas palavras consecutivas, como em um modelo de *bigramas*.


```python
# ✅ CONSTRUÇÃO GUIADA DO DICIONÁRIO — não altere nesta primeira execução.
proximas = {}
for frase in frases:
    palavras = palavras_da_frase(frase)
    for atual, seguinte in zip(palavras[:-1], palavras[1:]):
        if atual not in proximas:
            proximas[atual] = []
        proximas[atual].append(seguinte)

print("Quantidade de palavras-chave no dicionário:", len(proximas))
print("Depois de 'intelligence':", proximas.get("intelligence", []))
print("Depois de 'programming':", proximas.get("programming", []))
```

## Missão 3 · Calcule as probabilidades observadas

Já temos as possíveis continuações. Agora vamos descobrir **quantas vezes** cada uma aparece.

Se a palavra `can` aparece muitas vezes depois de `intelligence`, sua frequência relativa será maior. É essa proporção, calculada a partir do nosso pequeno conjunto de frases, que vamos observar. **Ainda não estamos usando um Transformer nem treinando uma rede neural.**


```python
# ✅ Contagem usando somente dicionários, laços e operações básicas.
def contar_ocorrencias(valores):
    resultado = {}
    for valor in valores:
        resultado[valor] = resultado.get(valor, 0) + 1
    return resultado

def tabela_proximas(palavra):
    opcoes = proximas.get(palavra.lower(), [])
    if not opcoes:
        print("Nenhuma continuação conhecida para:", palavra)
        return {}
    frequencias = contar_ocorrencias(opcoes)
    total = len(opcoes)
    for termo, qtd in sorted(frequencias.items(), key=lambda par: (-par[1], par[0])):
        print(f"{termo:<16} ocorrências={qtd:<2} proporção={qtd/total:.1%}")
    return frequencias

frequencias_intelligence = tabela_proximas("intelligence")
```

## Missão 4 · Faça a máquina prever

Vamos experimentar duas formas de escolher a próxima palavra:

- **`frequente`:** seleciona a continuação que mais apareceu nas nossas frases. Se houver empate, usa a ordem alfabética.
- **`sorteio`:** escolhe uma ocorrência aleatória da lista. Como uma palavra pode aparecer várias vezes, ela acaba tendo mais chances de ser escolhida.

Observem as saídas: o mesmo conjunto de dados pode produzir comportamentos diferentes dependendo da estratégia.


```python
import random

def prever_proxima(palavra, estrategia="frequente"):
    opcoes = proximas.get(palavra.lower(), [])
    if not opcoes:
        return None
    if estrategia == "sorteio":
        return random.choice(opcoes)
    if estrategia == "frequente":
        frequencias = contar_ocorrencias(opcoes)
        return sorted(frequencias, key=lambda termo: (-frequencias[termo], termo))[0]
    raise ValueError("Escolha 'frequente' ou 'sorteio'.")

print("Mais frequente:", prever_proxima("intelligence", "frequente"))
print("Sorteio 1:", prever_proxima("intelligence", "sorteio"))
print("Sorteio 2:", prever_proxima("intelligence", "sorteio"))
```

## Missão 5 · Gere uma sequência inteira

Até agora escolhemos apenas **uma** palavra. Vamos repetir esse processo várias vezes: a palavra sorteada passa a ser a referência para escolher a próxima, e assim construímos uma sequência.

Reparem se a frase continua fazendo sentido do começo ao fim. Quando ela fica estranha ou para antes da hora, isso também nos ajuda a entender os limites do algoritmo.


```python
# ✅ ESTA FUNÇÃO JÁ ESTÁ PRONTA.
def gerar_frase(inicio="Artificial intelligence", passos=8, estrategia="sorteio"):
    palavras = palavras_da_frase(inicio)
    if not palavras:
        return ""
    for _ in range(passos):
        nova = prever_proxima(palavras[-1], estrategia)
        if nova is None:
            break
        palavras.append(nova)
    return " ".join(palavras)

print("Frequente:", gerar_frase("Artificial intelligence", 10, "frequente"))
for n in range(3):
    print(f"Sorteio {n+1}:", gerar_frase("Artificial intelligence", 10, "sorteio"))
```

## Missão 6 · Avançando: use DUAS palavras de contexto

Vamos fazer uma mudança: em vez de olhar somente a última palavra, o programa vai considerar **as duas últimas**.

Para isso, o dicionário usará uma **tupla** como chave, por exemplo `('artificial', 'intelligence')`. Aqui juntamos três estruturas já estudadas em Programação: listas, dicionários e tuplas.

Executem as próximas células e observem se a continuação muda. Ter mais contexto pode ajudar, mas também pode fazer o programa parar quando não encontra aquele par no corpus.


```python
# ✅ NOVO DICIONÁRIO DE CONTEXTO COM DUAS PALAVRAS.
proximas_2 = {}
for frase in frases:
    palavras = palavras_da_frase(frase)
    for i in range(len(palavras) - 2):
        contexto = (palavras[i], palavras[i + 1])  # tupla: chave do dicionário
        seguinte = palavras[i + 2]
        proximas_2.setdefault(contexto, []).append(seguinte)

print("Após ('artificial', 'intelligence'):", proximas_2.get(("artificial", "intelligence")))
print("Após ('the', 'future'):", proximas_2.get(("the", "future")))
```

```python
def gerar_com_duas_palavras(inicio="Artificial intelligence", passos=8):
    palavras = palavras_da_frase(inicio)
    if len(palavras) < 2:
        return "Comece com pelo menos duas palavras."
    for _ in range(passos):
        contexto = (palavras[-2], palavras[-1])
        opcoes = proximas_2.get(contexto, [])
        if not opcoes:
            break
        palavras.append(random.choice(opcoes))
    return " ".join(palavras)

print("Uma palavra de contexto:", gerar_frase("Artificial intelligence", 8))
print("Duas palavras de contexto:", gerar_com_duas_palavras("Artificial intelligence", 8))
```

### Dois parâmetros para entender antes do teste

**1. `temperature` — como o modelo distribui as chances**

Antes de escolher um novo token (que pode ser uma palavra ou parte dela), o modelo calcula pontuações para as possibilidades. A temperatura altera a concentração das probabilidades usadas no sorteio:

- **`temperature=1.0`**: mantém a distribuição de probabilidades original.
- **`temperature=0.3`**: favorece ainda mais as opções que já eram mais prováveis; o resultado tende a variar menos.
- **Acima de `1.0`**, como `1.5`: aumenta relativamente a chance de alternativas menos prováveis; pode aumentar a diversidade, mas também produzir texto menos coerente.

**Quais são os limites?** Neste uso da biblioteca Transformers, a temperatura para amostragem precisa ser **maior que zero**. Não há um limite máximo universal de `1` ou `2`. Temperaturas extremamente baixas concentram muito as probabilidades; temperaturas muito altas espalham demais as chances. **Não use `temperature=0` com `do_sample=True` neste Colab.** Outras ferramentas podem tratar zero de maneira especial.

**Exemplo apenas para entender:** imagine três opções com probabilidades de 70%, 20% e 10%. Com temperatura `1.0`, elas ficam como estão. Com `0.3`, a primeira opção passa a concentrar aproximadamente 98% das chances. Isso **não** torna o modelo mais inteligente ou mais correto.

**2. `do_sample` — sorteio ligado ou desligado**

- **`do_sample=True`**: o modelo sorteia o próximo token conforme as probabilidades. As respostas podem variar mesmo com a mesma frase inicial.
- **`do_sample=False`**: não faz esse sorteio. No modo simples com `num_beams=1`, escolhe a opção mais provável a cada passo (*greedy decoding*). Nessa configuração, a temperatura não atua na escolha e deve ser omitida.

**O que vamos testar:** dois inícios de frase em inglês, com `do_sample=True`, comparando `temperature=0.3` e `1.0`. Observem repetição, diversidade e coerência. **Uma saída diferente não é automaticamente uma saída melhor.** Se o modelo não carregar, a comparação conceitual continua valendo.


```python
# ✏️ O PROFESSOR LIBERA ESTA ETAPA MUDANDO False PARA True.
RUN_TRANSFORMER = True

if not RUN_TRANSFORMER:
    print("DEMO OPCIONAL DESATIVADA: o laboratório básico funciona sem downloads.")
else:
    try:
        from transformers import pipeline
        gerador = pipeline("text-generation", model="distilbert/distilgpt2", device=-1)
        print("DistilGPT2 carregado com sucesso.")
    except Exception as erro:
        RUN_TRANSFORMER = False
        print("Não foi possível carregar o Transformer:", type(erro).__name__, str(erro)[:260])
        print("Registre o impedimento e continue a atividade principal.")
```

```python
# ✅ Comparação de variação: a temperatura só é usada quando do_sample=True.
# Os resultados dependem do modelo, ambiente e sorteio — não são garantidos.
prefixos = ["Artificial intelligence can", "The future of technology"]
if RUN_TRANSFORMER:
    for inicio_texto in prefixos:
        print("\nPREFIXO:", inicio_texto)
        for t in (0.3, 1.0):
            saida = gerador(
                inicio_texto, max_new_tokens=20, do_sample=True,
                temperature=t, pad_token_id=gerador.tokenizer.eos_token_id,
                truncation=True
            )[0]["generated_text"]
            print(f"Temperatura {t}:", saida)
else:
    print("Etapa Transformer não executada; registre isso no relatório.")
```
