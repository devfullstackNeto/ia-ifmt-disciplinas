
# Classe de Negócio


```python
class Aluno:

    def __init__(self, nome, email):
        self.nome = nome
        self.email = email

    def __str__(self):
        return f"{self.nome} - {self.email}"
```


# Conexão com o Banco
# Classe DAO

```python
import sqlite3

class AlunoDAO:

    def __init__(self):

        self.conexao = sqlite3.connect("escola.db")

        self.cursor = self.conexao.cursor()

        self.cursor.execute("""
        CREATE TABLE IF NOT EXISTS aluno(
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            nome TEXT NOT NULL,
            email TEXT NOT NULL
        )
        """)

        self.conexao.commit()

    def inserir(self, aluno):

        sql = """
        INSERT INTO aluno(nome,email)
        VALUES(?,?)
        """

        self.cursor.execute(
            sql,
            (aluno.nome, aluno.email)
        )

        self.conexao.commit()

    def listar(self):

      self.cursor.execute(
          "SELECT * FROM aluno"
      )

      return self.cursor.fetchall()

    def buscar_por_id(self, id):

      self.cursor.execute(
          "SELECT * FROM aluno WHERE id=?",
          (id,)
      )

      return self.cursor.fetchone()

    def atualizar(self,id,nome,email):

      self.cursor.execute(
          """
          UPDATE aluno
          SET nome=?, email=?
          WHERE id=?
          """,
          (nome,email,id)
      )

      self.conexao.commit()

    def excluir(self,id):

      self.cursor.execute(
          """
          DELETE FROM aluno
          WHERE id=?
          """,
          (id,)
      )

      self.conexao.commit()
```

# CREATE

```python
a = Aluno(
    "Carlos",
    "carlos@ifmt.edu.br"
)

dao = AlunoDAO()
dao.inserir(a)
```

```python
for aluno in dao.listar():
    print(aluno)
```

```python
dao.atualizar(
    2,
    "Carlos Gracioli",
    "gracioli@ifmt.edu.br"
)
```

```python
resultado = dao.buscar_por_id(4)

print(resultado)
```

```python
dao.excluir(3)
```
