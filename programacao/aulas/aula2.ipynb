```python
numeros = [1, 2, 3, 4, 5]
pares = list(filter(lambda x: x % 2 == 0, numeros))
quadrados = list(map(lambda x: x ** 2, pares))
print(quadrados)
```

```python
numeros = [1, 2, 3, 4, 5]
quadrados = []
for item in numeros:
  if item % 2 == 0:
    quadrados.append(item)
    print(item ** 2)
```

```python
numeros = [1, 2, 3, 4, 5]
quadrados = [x**2 for x in numeros if x%2==0]
print(quadrados)
```
